### Exrtracting the year,month and day

````python
# extracting year
data["order_year"]= data["order_date"].dt.year
data["order_month"]= data["order_date"].dt.month_name()
data["order_day"]= data["order_date"].dt.day_name()
data.head()
````

````python
order_date = data['order_date']
daliver_date = data['delivery_date']
last_login = data['last_login']
subscription_start = data['subscription_start']

dates_columns = {
'year of order':order_date, 
'year of delivery':daliver_date ,
'year of last_login':last_login ,
'yaer of subscription_start':subscription_start
}

for index,year in dates_columns.items():
    data[index] = year.dt.year.astype('Int64')
data   
````

````python
date_columns= [
    'order_date',
    'delivery_date',
    'last_login',
    'subscription_start'
]

for col in date_columns:
    data[f'{col}_month'] = data[col].dt.month
data    
````

````python
date_cols = [
    'order_date',
    'delivery_date',
    'last_login',
    'subscription_start'
]

for col in date_cols:
    data[f'{col}_day'] = data[col].dt.day_name()
data
````

````python
dates_columns = {
'order_quarter':order_date, 
'delivery_quarter':daliver_date ,
'last_login-quarter':last_login ,
'subscription_start_quarter':subscription_start
}

for name,quarter in dates_columns.items():
    data[name] = quarter.dt.quarter.astype('Int64')
data   
````

## Main Types of errors
- SyntaxError
- RuntimeError
- LogicalErrors(semantic Errors)
### SyntaxError
Syntax errors happen when the code violates the grammatical rules of the Python language
- The Cause: The Python interpreter checks syntax before running the script. If it finds a typo or improper layout, the program will crash immediately without running any code.
- Example: Forgetting a colon at the end of an if statement or leaving a parenthesis unclosed
### RuntimeError
Runtime errors occur while the program is actively executing.
- The Cause: The syntax is perfectly valid, but the computer hits an impossible instruction or operation given the current state of the environment. 
   These are frequently referred to as Python Built-In Exceptions.
   
```
   result = 10 / 0  # ZeroDivisionError: division by zero

```

- **Example**: Forcing -Python to divide by zero (ZeroDivisionError) or using a variable name that hasn't been created yet (NameError).

### LogicalErrors(semantic Errors)
Logical errors mean the program runs entirely from start to finish without generating any official crash messages or exceptions, but it produces the wrong output
- The Cause: The flaw lies strictly in the programmer’s logic, algorithm, or calculations.
-Example: Attempting to calculate the average of two numbers but forgetting to use proper math grouping symbols.

```
# Intending to average 4 and 6. Expected result is 5.
average = 4 + 6 / 2  # Result is 7.0 due to order of operations. 
```

### errors
- errors='coerce' means that any invalid data or parsing errors will be forced into a missing value placeholder instead of crashing your script.
- errors='raise' (Default): Stops execution immediately and throws an error if it finds bad data.
- errors='ignore': Leaves the invalid data exactly as it is and returns the original input unchanged
