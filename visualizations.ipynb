## STEP 1: Import Libraries

```python
import pandas as pd  
import seaborn as sns 
import matplotlib.pyplot as plt 

# If you want to view all the columns with truncation
pd.set_option('display.max_columns', None)  

# If you want to view all the rows
# pd.set_option('display.max_rows', None) # Use it cautiously as it might make your pc to be slow
```

## STEP 2: Load the dataset

```python
housing_df = pd.read_csv("nairobi_housing_statistics_dataset.csv")

housing_df.head()
```

### Visualing using matplotlib and seaborn  
__a). CountPlot__

```python
# Visualizing properties by property type
plt.figure(figsize=(6, 3))  

sns.countplot(data=housing_df, x="property_type", palette='dark')
plt.ylabel("Count/Number of Properties")
plt.xticks(rotation=45)
plt.title("Count per property type")  
plt.xlabel("Property Type")

plt.show()
```

__b). Bar Plot__  
Use barplots when comparing aggregated values, such as average rent, average occupacy rate etc.

```python
# Average monthly rent by property type  
plt.figure(figsize=(6, 3))

sns.barplot(data=housing_df, x="property_type", y="monthly_rent_kes", estimator="mean", palette='bright')
plt.title("Average monthly rent by property type")
plt.xlabel("Property type")
plt.ylabel("Average monthly rent")  
plt.show()
```

```python
# Average satisfaction score by property type  
avg_satisfaction_by_prop = housing_df.groupby("property_type")["satisfaction_score"].mean().sort_values(ascending=False).reset_index()  

# Plot  
plt.figure(figsize=(6, 3))  

sns.barplot(data=avg_satisfaction_by_prop, x="property_type", y="satisfaction_score")  
plt.title("Average satisfaction score by property type")
plt.xlabel("Property Type")
plt.ylabel("Average Satisfaction score")  
plt.show()
```

__c). Histograms__  
We use histograms to show distribution of a numerical and categorical columns  
_Numerical_

```python
# Plot the distribution of the monthly rent KES  
plt.figure(figsize=(6, 3)) 

sns.histplot(data=housing_df, x="monthly_rent_kes", bins=30, kde=True)
plt.xlabel("Monthly Rent")
plt.show()
```

```python
# Plotting distribution of the floor size  
plt.figure(figsize=(6, 3))

sns.histplot(data=housing_df, x="floor_size_sqm", bins=30, kde=True)
plt.xlabel("Floor size in SQM")
plt.show()
```

__d). BoxPlots__  
Use boxplot to compare spread, median and outliers.

__e). Scatter plots__  
Use scatter plots to show the relationship between two numerical variables.

```python
# Floor size vs monthly rent  
plt.figure(figsize=(6, 3))  

sns.scatterplot(data=housing_df, x="floor_size_sqm", y="monthly_rent_kes")  
plt.title("Relationship between floor size and monthly rent")  
plt.xlabel("Floor Size")
plt.ylabel("Monthly Rent")
plt.show()
```

```python
# Floor size vs monthly rent  
plt.figure(figsize=(6, 3))  

sns.scatterplot(data=housing_df, x="monthly_rent_kes", y="monthly_rent_kes")  
plt.title("Relationship between monthly rent and monthly rent")  
plt.xlabel("Monthly Rent")
plt.ylabel("Monthly Rent")
plt.show()
```

```python
# Plotting relationship between bedrooms and bathrooms  
plt.figure(figsize=(6, 3))  

sns.scatterplot(data=housing_df, x="bathrooms", y="bedrooms")
plt.title("Bathrooms vs Bedrooms")
plt.xlabel("Bathrooms")
plt.ylabel("Bedrooms")
plt.show()
```

__f). Line Plots__  
Line plots are best for showing trends.  

```python
plt.figure(figsize=(6, 3))  

sns.lineplot(data=housing_df, x="bathrooms", y="bedrooms")
plt.title("Bathrooms vs Bedrooms")
plt.xlabel("Bathrooms")
plt.ylabel("Bedrooms")
plt.show()
```

```python
# Average rent trend from 2024 to 2025  
rent_trend = housing_df[["rent_2024", "rent_2025"]].mean().reset_index()
 

rent_trend.columns = ["year","average_rent"]

rent_trend  

plt.figure(figsize=(6, 3))

sns.lineplot(data=rent_trend, x="year", y="average_rent", marker="o")  
plt.title("Average rent trend from 2024 to 2025")
plt.xlabel("Year")
plt.ylabel("Avg rent")
plt.show()
```

__g). Pie Charts__    
Seaborn does not have directly the functionality to create a piechart, so we use matplotlib

```python
# Plotting a pie chart for the column vacancy status   
#
# 
explode = (0, 0.1) 

vacancy_counts = housing_df["vacancy_status"].value_counts()

plt.figure(figsize=(6, 3))  

plt.pie(vacancy_counts, explode=explode, labels=vacancy_counts.index, autopct="%1.1f%%")  
plt.title("Vacancy Distribution")
plt.show()
```

```python
# Plotting furnishing status  
furnishing_counts = housing_df["furnishing"].value_counts()

plt.figure(figsize=(6, 6))

exploded = (0.05, 0.05, 0.05)

plt.pie(furnishing_counts, explode = exploded, labels=furnishing_counts.index, autopct="%1.1f%%")
plt.title("Distribution of furnishing status")
plt.show()
```

__h). Heatmaps__   
We use heatmaps to show relationships between numerical variables.   
_Interpreting correlation_    

-1 to -0.5 > Strong negative correlation  
-0.5 to 0 > Weak negative correlation   
0 to 0.5 > weak positive correlation   
0.5 to 1 > strong positive  

```python
# Select numerical columns
numerical_columns = housing_df.select_dtypes(include=["int", "float"]).columns.to_list()

numerical_columns
```

```python
numerical_columns.remove("property_id")

numerical_columns
```

```python
# Correlation analysis  
correlation  = housing_df[numerical_columns].corr()  

plt.figure(figsize=(10, 6))

sns.heatmap(correlation, annot=True, fmt=".2f")
plt.title("Correlation Heatmap for Numerical Variables")
plt.show()
```

### Univariate, Bivariate, Multivariate Analysis  
__a). Univariate__  
- Univariate analysis means analyzing one variable/column/feature at a time.   
It helps us understand:  
1. Distribution  
2. Frequency  
3. Spread  
4. Outliers  

_i) Numerical Columns_

```python
# What is the distribution pof monthly rent   
plt.figure(figsize=(6, 3))  

sns.histplot(data=housing_df, x="monthly_rent_kes",
             bins=30, kde=True)

plt.title("Distribution of monthly rent")  
plt.xlabel("Monthly Rent KES")
plt.ylabel("Number of properties")
plt.show()
```

_ii) Categorical data_

```python
# Counts of property type
plt.figure(figsize=(6, 3))  

sns.countplot(data=housing_df, x="property_type", palette="dark")

plt.title("Number of Properties by Property Type")  
plt.xlabel("Property Type")  
plt.ylabel("Count")
plt.show()
```

__b). Bivariate  Analysis__
- Bivariate analysis means analyzing the relationship between two variables. 
- It helps us understand:  
    1. Relationships  
    2. Comparisons.  
    3. Association.  
    4. Correlation.  
    5. Difference between groups.    

_i) Numerical vs Numerical_

```python
# Floor size vs Monthly rent  
plt.figure(figsize=(6, 3))  

sns.scatterplot(housing_df, x="floor_size_sqm", y="monthly_rent_kes")

plt.title("Monthly rent vs Floor Size")
plt.xlabel("Floor Size sqm")
plt.ylabel("Monthly rent kes")
plt.show()
```

_ii) Numerical vs Categorical_  


```python
# Monthly rent vs property type  
plt.figure(figsize=(6, 3))  

sns.barplot(housing_df, x="property_type", y="monthly_rent_kes")
plt.title("Monthly rent by property type")
plt.xlabel("Property Type")
plt.ylabel("Monthly Rent")  
plt.show()
```

_iii) Categorical vs Categorical_  


```python
# Property type vs vacancy Status  

plt.figure(figsize=(6, 3))

sns.countplot(housing_df, x="property_type", hue="vacancy_status")  
plt.title("Vacancy by property type")
plt.xlabel("Property Type")
plt.ylabel("Number of properties")   
plt.show()
```

__c). Multivariate__  
Multivariate analysis means analyzing three or more variables at the same time.  

it helps us understand more complex relationshhips.  


```python
# Average Rent by property type and furnishing   
plt.figure(figsize=(6,3))

sns.barplot(housing_df, x="property_type", y="monthly_rent_kes", hue="furnishing", estimator="mean")  
plt.title("Average rent by property type and furnishing")
plt.xlabel("Property Type")
plt.ylabel("Average Monthly Rent in KES")
plt.legend(title="Furnishing")   
plt.show()
```
